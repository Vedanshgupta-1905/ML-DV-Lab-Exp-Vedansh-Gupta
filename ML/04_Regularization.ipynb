# Experiment 4: Regularization
**Aim:** Implement and evaluate Regularization using Python and scikit-learn.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import Ridge,Lasso
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
X,y=load_diabetes(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
for estimator in [Ridge(alpha=1),Lasso(alpha=1)]:
    model=make_pipeline(StandardScaler(),estimator)
    model.fit(X_train,y_train)
    pred=model.predict(X_test)
    print(type(estimator).__name__,"MSE:",mean_squared_error(y_test,pred),"R2:",r2_score(y_test,pred))
```
