# Experiment 9: K Means Clustering
**Aim:** Implement and evaluate K Means Clustering using Python and scikit-learn.

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
X,y=load_iris(return_X_y=True)
Z=StandardScaler().fit_transform(X)
model=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=model.fit_predict(Z)
print("Silhouette score:",silhouette_score(Z,labels))
plt.scatter(Z[:,0],Z[:,1],c=labels,cmap="viridis")
plt.xlabel("Standardized sepal length");plt.ylabel("Standardized sepal width");plt.show()
```
